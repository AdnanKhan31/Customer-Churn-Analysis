```python
## Importing Panda Libraries
import pandas as pd
from sqlalchemy import create_engine
```

```python
## Load Data and Show Total Rows and Columns
df = pd.read_csv(r"C:\Users\Adnan khan\Downloads\WA_Fn-UseC_-Telco-Customer-Churn.csv")

print(df.shape)
print(df.head())
```

```python
## Summary of Data
df.info()
```

```python
## Finding Missing Values From DataSet
df.isnull().sum()
```

```python
## Finding Duplicate Values From DataSet
df.duplicated().sum()
```

```python
## Remove Extra Spaces
df = df.apply(lambda x: x.str.strip() if x.dtype == "str" or x.dtype == "object" else x)
```

```python
## Finding Unique Values From All Columns
for col in df.select_dtypes(include="str").columns:
    print(col, ":", df[col].unique())
```

```python
## Use this to check if Extra Space Exists?
for col in df.select_dtypes(include="str").columns:
    print(col, df[col].ne(df[col].str.strip()).sum())
```

```python
## Finding All Unique Values in Particular Columns
df["TotalCharges"].unique()
```

```python
## Converting Text Column to Numerical Column 
cols = ["TotalCharges"]
for c in cols: df[c] = pd.to_numeric(df[c],errors="coerce")
```

```python
## Checking if the Column in Converted 
df["TotalCharges"].dtype
```

```python
df.isnull().sum()
```

```python
# Checking If Tebure is New Customer 
df.loc[df["TotalCharges"].isna(), "tenure"]
```

```python
## Making the Missing Values of Tenure To 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df.isnull().sum()
```

```python
## Statistical Summary of Numerical Data
df.describe()
```

```python
## Finding Negative Values in Numerical Columns
df.select_dtypes(include="number").lt(0).sum()
```

```python
## Finding a Range for Numerical Columns
for col in df.select_dtypes(include="number").columns:
    print(col, "Min:", df[col].min(), "| Max:", df[col].max())
```

```python
## Finding Overall Churn Value
df["Churn"].value_counts()
```

```python
## Overall Churn Percentage
churn_percentage = (df["Churn"].value_counts(normalize=True) * 100)
print(churn_percentage)
```

```python
## Contract Type Customer Churn
pd.crosstab(df["Contract"], df["Churn"])
```

```python
## Percentage of Contract Type Customer Churn
pd.crosstab(df["Contract"], df["Churn"], normalize="index") * 100
```

```python
## Tenure Group Churn
pd.crosstab(df["tenure"] // 12, df["Churn"], normalize="index") * 100
```

```python
## Percentage of Tenure Group Churn
pd.crosstab(pd.cut(df["MonthlyCharges"], bins=3), df["Churn"], normalize="index") * 100
```

```python
## Churn By internet Services
pd.crosstab(df["InternetService"], df["Churn"], normalize="index") * 100
```

```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\Adnan khan\Downloads\WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
).fillna(0)

print(df.shape)
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:mohammedadnankhan@localhost:5432/Telecom_project"
)

with engine.connect() as connection:
    print("Connected successfully!")
```

```python
df.to_sql(
    "Telecom_project",
    con=engine,
    if_exists="replace",
    index=False
)

print("Dataset imported successfully!")
```
